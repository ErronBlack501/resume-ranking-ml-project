```python
# Distribution des scores

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.hist(df["matched_score"], bins=30)
plt.xlabel("Matched score")
plt.ylabel("Nombre de lignes")
plt.title("Distribution de matched_score")
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10, 8))
order = df.groupby("job_position_name")["matched_score"].median().sort_values().index
sns.boxplot(data=df, y="job_position_name", x="matched_score", order=order)
plt.title("Distribution du matched_score par poste")
plt.tight_layout()
plt.show()
```

```python
df["career_objective_len"] = df["career_objective"].fillna("").str.len()
df["skills_count"] = df["skills"].fillna("[]").apply(lambda x: x.count(",") + 1 if x != "[]" else 0)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(df["skills_count"], df["matched_score"], alpha=0.1, s=5)
axes[0].set_xlabel("Nombre de skills"); axes[0].set_ylabel("matched_score")

axes[1].scatter(df["career_objective_len"], df["matched_score"], alpha=0.1, s=5)
axes[1].set_xlabel("Longueur career_objective"); axes[1].set_ylabel("matched_score")
plt.tight_layout()
plt.show()
```

```python
import ast

def parse_skills(x):
    try:
        return set(s.strip().lower() for s in ast.literal_eval(x))
    except Exception:
        return set()

def parse_required(x):
    if pd.isna(x):
        return set()
    return set(s.strip().lower() for s in x.split("\n") if s.strip())

df["skills_set"] = df["skills"].apply(parse_skills)
df["required_set"] = df["skills_required"].apply(parse_required)
df["skills_overlap"] = df.apply(
    lambda r: len(r["skills_set"] & r["required_set"]), axis=1
)

plt.figure(figsize=(8, 5))
sns.boxplot(x="skills_overlap", y="matched_score", data=df[df["skills_overlap"] <= 5])
plt.title("matched_score selon le nombre de skills en commun")
plt.show()
```
